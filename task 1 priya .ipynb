```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Price (₹)': [20, 25, 30, 35, 40],
    'Bottles Sold': [120, 110, 90, 80, 70]
}
df = pd.DataFrame(data)
df
```

```python
df['Revenue (₹)'] = df['Price (₹)'] * df['Bottles Sold']
df
```

```python
max_revenue = df['Revenue (₹)'].max()
best_price = df[df['Revenue (₹)'] == max_revenue]['Price (₹)'].values[0]

print(f"HIGHEST REVENUE: ₹{max_revenue}")
print(f"BEST PRICE: ₹{best_price}")
print("\n")
```

```python
plt.figure(figsize=(10, 6))
bars = plt.bar(df['Price (₹)'], df['Revenue (₹)'], 
               color=['#FF6B6B', '#4CAF50', '#FFC107', '#2196F3', '#9C27B0'],
               edgecolor='black')

for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 50,
             f'₹{int(height)}', ha='center', va='bottom', fontweight='bold')
for bar in bars:
    if bar.get_height() == max_revenue:
        bar.set_color('#FF0000')
        bar.set_linewidth(3)

plt.title('Juice Shop: Price vs Revenue Analysis', fontsize=16, fontweight='bold')
plt.xlabel('Selling Price (₹)', fontsize=13, fontweight='bold')
plt.ylabel('Total Revenue (₹)', fontsize=13, fontweight='bold')
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.show()
```
